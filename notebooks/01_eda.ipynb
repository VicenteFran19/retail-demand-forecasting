# EDA - Retail Demand Forecasting

Analisis exploratorio del dataset sintetico antes de modelar.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet('../data/ventas_sinteticas.parquet')
df.head()
```

```python
df.groupby('fecha')['unidades_vendidas'].sum().plot(figsize=(14,4), title='Demanda total diaria (todas las tiendas/productos)')
plt.ylabel('Unidades vendidas')
plt.show()
```

```python
df.groupby(df['fecha'].dt.dayofweek)['unidades_vendidas'].mean().plot(kind='bar', title='Demanda promedio por dia de semana (0=lunes)')
plt.show()
```

```python
df.groupby(df['fecha'].dt.month)['unidades_vendidas'].mean().plot(kind='bar', title='Demanda promedio por mes (estacionalidad anual)')
plt.show()
```

```python
print('Dias con quiebre de stock:', df['hubo_quiebre_stock'].sum(), f"({df['hubo_quiebre_stock'].mean()*100:.2f}% del total)")
print('Dias en promocion:', df['en_promocion'].sum(), f"({df['en_promocion'].mean()*100:.2f}% del total)")
```
